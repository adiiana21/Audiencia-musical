## Etapa 2. Preprocesamiento de datos <a id='data_preprocessing'></a>

El objetivo aquí es preparar los datos para que sean analizados.
El primer paso es resolver cualquier problema con los encabezados. Luego podemos avanzar a los valores ausentes y duplicados.

Se corrige el formato en los encabezados de la tabla.


### Estilo del encabezado <a id='header_style'></a>
Se muestra los encabezados de la tabla (los nombres de las columnas):

```python
# Muestra los nombres de las columnas
df.columns
```

Se cambia los encabezados de la tabla de acuerdo con las reglas del buen estilo:
* Todos los caracteres deben ser minúsculas.
* Elimina los espacios.
* Si el nombre tiene varias palabras, utiliza snake_case.

```python
#bucle for para iterar sobre los nombres de las columnas y poner todos los caracteres en minúsculas
new_columns=[] # crea una nueva lista vacia
for old_column in df.columns: #itera sobre las columnas
    new_name= old_column.lower() # metodo para cambiar a minusculas
    new_columns.append(new_name) # llena la nueva lista con los valores nuevos
df.columns=new_columns # define las nuevas comunas con los valores de la lista
print(df.columns)
```

Utilizando el mismo método, se eliminan los espacios al principio y al final de los nombres de las columnas y se imprimen los nombres de las columnas nuevamente:

```python
# Bucle en los encabezados eliminando los espacios
new_columns=[]
for old_column in df.columns:
    new_name= old_column.lower().strip()
    new_columns.append(new_name)
df.columns=new_columns
print(df.columns)
```

Se aplica la regla de snake_case a la columna `userid`. Debe ser `user_id`. El nombre de esta columna se cambia y se muestran los nombres de todas las columnas al final.

```python
# Cambiar el nombre de la columna "userid"
new_columns=[]
for old_column in df.columns:
    new_name= old_column.lower().strip().replace('userid','user_id')
    new_columns.append(new_name)
df.columns=new_columns
print(df.columns)
```

```python
# Comprobar el resultado: la lista de encabezados
print(df.columns)
```

### Duplicados <a id='duplicates'></a>
Se encuentra el número de duplicados explícitos en la tabla. Se aplican dos métodos en secuencia para obtener la cantidad de duplicados explícitos. Posteriormente se eliminan los duplicados.

```python
# Contar duplicados explícitos
print(df.duplicated().sum())
```

```python
# Eliminar duplicados explícitos
df.drop_duplicates(inplace=True)
```

```python
# Comprobar de nuevo si hay duplicados
print(df.duplicated().sum())
```

Se eliminan los duplicados implícitos en la columna `genre`. Por ejemplo, el nombre de un género se puede escribir de varias formas. Dichos errores también pueden afectar al resultado.

Para hacerlo, primero mostremos una lista de nombres de género únicos, ordenados en orden alfabético. Para ello:
* Se extrae la columna `genre` del DataFrame.
* Se llama al método que devolverá todos los valores únicos en la columna extraída.


```python
# Inspeccionar los nombres de géneros únicos en genre.
pr=sorted(df['genre'].unique())
print(pr)
```

De la lista se encontran duplicados implícitos del género `hiphop`. Estos son nombres escritos incorrectamente o nombres alternativos para el mismo género. Como por ejemplo los siguientes duplicados implícitos:
* `hip`
* `hop`
* `hip-hop`

Para cambiar los nombres, se crea una función llamada `replace_wrong_genres()` con dos parámetros:
* `wrong_genres=`: una lista que contiene todos los valores a reemplazar.
* `correct_genre=`: un string que se utiliza como reemplazo.

Como resultado, la función corrige los nombres en la columna `'genre'` de la tabla `df`, remplaza cada valor de la lista `wrong_genres` por el valor en `correct_genre`.

Dentro del cuerpo de la función, se utiliza un bucle `'for'` para iterar sobre la lista de géneros incorrectos, se extrae la columna `'genre'` y aplica el método `replace` para hacer correcciones.

```python
# Función para reemplazar duplicados implícitos
def replace_wrong_genres(wrong_genres, correct_genre):
    for wrong_value in wrong_genres: #itera sobre los valores dentro de wrong_genres
        df['genre']=df['genre'].replace(wrong_value, correct_genre) #reemplaza los valores incorrectos por los correctos
    return df
```

Ahora, se llama a `replace_wrong_genres()` y se le pasa los argumentos para que retire los duplicados implícitos (`hip`, `hop` y `hip-hop`) y los reemplace por `hiphop`:

```python
# Eliminar duplicados implícitos
wg=['hip', 'hop', 'hip-hop'] #generos incorrectos
cg='hiphop' #genero correcto
df=replace_wrong_genres(wg,cg) #funcion para reemplazar en el dataset df
print(df)
```

Se comprueba que los nombres duplicados han sido eliminados. Se muestra la lista de valores únicos de la columna `'genre'` una vez más:

```python
# Comprobación de duplicados implícitos
print(df['genre'].unique()) #valores unicos
```

## Etapa 3. Prueba de hipótesis <a id='hypothesis'></a>

### Hipótesis: comparar el comportamiento del usuario o la usuaria en las dos ciudades <a id='activity'></a>

La hipótesis afirma que existen diferencias en la forma en que los usuarios y las usuarias de Springfield y Shelbyville consumen música. Para comprobar esto, se usa los datos de tres días de la semana: lunes, miércoles y viernes.

* Se agrupan a los usuarios y las usuarias por ciudad.
* Se compara el número de canciones que cada grupo reprodujo el lunes, el miércoles y el viernes.


Cada cálculo de ciudad se realiza por separado por separado.

El primer paso es evaluar la actividad del usuario en cada ciudad.  Las etapas son dividir-aplicar-combinar,  el objetivo ahora es agrupar los datos por ciudad, aplicar el método apropiado para contar durante la etapa de aplicación y luego encontrar la cantidad de canciones reproducidas en cada grupo especificando la columna para obtener el recuento.

Para evaluar la actividad de los usuarios y las usuarias en cada ciudad, se agrupan los datos por ciudad y se encuentra la cantidad de canciones reproducidas en cada grupo.



```python
# Contar las canciones reproducidas en cada ciudad
track_num=df.groupby('city')['track'].count()
print(track_num)
```

Al clasificar por ciudad, observamos que en Springfield se reprodujeron más del doble de canciones que en Shelbyville.

Ahora se agrupan los datos por día de la semana y se encuentra el número de canciones reproducidas el lunes, miércoles y viernes. Se utiliza el mismo método de antes, pero ahora una agrupación diferente.


```python
# Calcular las canciones reproducidas en cada uno de los tres días
by_day=df.groupby('day')['track'].count()
print(by_day)
```

Al clasificar por dia de la semana, se observa que el viernes es el día con más reproducciones de canciones, seguido del lunes, mientras que el miercoles es el dia con menor canciones reproducidas. 

Ya se tienen las agrupaciones por ciudad o día. Ahora necesitas escribir una función que pueda contar entradas según ambos criterios simultáneamente.

Se crea la función `number_tracks()` para calcular el número de canciones reproducidas en un determinado día **y** ciudad. La función acepta los parámetros:

- `day`: un día de la semana para filtrar. Por ejemplo, `'Monday'` (lunes).
- `city`: una ciudad para filtrar. Por ejemplo, `'Springfield'`.

Dentro de la función, se aplica un filtrado consecutivo con indexación lógica.

Primero filtra los datos por día y luego filtra la tabla resultante por ciudad.

Después de filtrar los datos por dos criterios, se cuenta el número de valores de la columna 'user_id' en la tabla resultante. Este recuento representa el número de entradas que se buscan. Por ultimo se guarda el resultado en una nueva variable y se devuélve desde la función.

```python
# Declara la función number_tracks() con dos parámetros: day= y city=.
def number_tracks(day, city):

    # Almacena las filas del DataFrame donde el valor en la columna 'day' es igual al parámetro day=
    df_days=df[df['day']==day]
    # Filtra las filas donde el valor en la columna 'city' es igual al parámetro city=
    df_city=df_days[df_days['city']==city]
    # Extrae la columna 'user_id' de la tabla filtrada y aplica el método count()
    us_ft_cd=df_city['user_id'].count()
    # Devolve el número de valores de la columna 'user_id'
    return us_ft_cd
```

Se llama a `number_tracks()` seis veces, donde los valores de los parámetros cambian para recuperar los datos de ambas ciudades para cada uno de los tres días.

```python
# El número de canciones reproducidas en Springfield el lunes
print(number_tracks('Monday', 'Springfield'))
```

```python
# El número de canciones reproducidas en Shelbyville el lunes
print(number_tracks('Monday', 'Shelbyville'))
```

```python
# El número de canciones reproducidas en Springfield el miércoles
print(number_tracks('Wednesday', 'Springfield'))
```

```python
# El número de canciones reproducidas en Shelbyville el miércoles
print(number_tracks('Wednesday', 'Shelbyville'))
```

```python
# El número de canciones reproducidas en Springfield el viernes
print(number_tracks('Friday', 'Springfield'))
```

```python
# El número de canciones reproducidas en Shelbyville el viernes
print(number_tracks('Friday', 'Shelbyville'))
```

**Observaciones**

La hipotesis planteada es: La actividad de los usuarios y las usuarias difiere según el día de la semana y dependiendo de la ciudad.

Analizando los datos y clasificandolos por dia y por ciudad podemos observar que la actividad de los usuarios si difiere en cuanto a dia de la semana y dependiendo de la ciudad ya que en el caso de Springfield el dia con menor usuarios es el miercoles mientras que para Shelbyville es el miercoles el dia de mayor cantidad de usuarios. Por otro lado la cantidad de usuarios para los dias lunes y viernes es muy similar en entre esos dias para ambas ciudades, sin embargo en Shelbyville es menor a su dia de maxima cantidad de usuarios y para Springfield estos son los dias con mayor cantidad de usuarios.
